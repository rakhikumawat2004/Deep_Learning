```python
# POS TAGGING IN NLP
```

```python
# pos taggiing means part of speech tagging
# it is an nlp technique where each word in a sentence is assigned
# its grammatical roles such as:
# Noun
# Verb
# Adjective
# Pronoun
# preposition
# conjuction
```

```python
# Why POS TAGGING is important?
# # POS TAGGINGis used in:
# Text classification
# chatbots
# machine translation
# name entity recognition
# sentiment analysis
# speech recognition
# grammer checking
```

```python
# common pos tags
```

```python
# |pos tags | meaning         |
# | NN      | Noun            |
# |NNS      | plural noun     |
# |NNP      | proper noun     |
# |VB       | verb            |
# |VBD      | verb post tense |
# |VBG      | verb gerund     |
# |JJ       | adjective       |
# |RB       | adverb          |
# |PRP      | pronoun         |
# |IN       | preposition     |
```

```python
# working process of pos tagging
# 1 sentence is divided into tokens
# 2 NlP model checks grammer/context
# 3 appropriate pos tag is assigned to each word
```

```python
import spacy

#load english model
nlp=spacy.load('en_core_web_sm')

#input text
doc=nlp("Shweta is learning NLP")

# pos tagging
for token in doc:
  print(token.text,"------->",token.pos_)
```

```python
#example 2
nlp=spacy.load("en_core_web_sm")
text="shweta play cricket"
doc=nlp(text)
for token in doc:
  print(token.text,"------->",token.pos_)
```

```python
#example 2
nlp=spacy.load("en_core_web_sm")
text="shweta is learning nlp"
doc=nlp(text)
for token in doc:
  print("word:",token.text,
        "pos:",token.pos_,
        "tag:",token.tag_)
```
